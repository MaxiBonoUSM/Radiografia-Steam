## 3. Mapeo X, Y y T

Se deja documentado el rol que cumplira cada variable en el analisis posterior. Los nombres exactos de columna
se ajustaran una vez confirmada la version del dataset.

- X (informacion disponible): genero, precio, idiomas soportados, soporte de sistemas operativos,
  funcionalidades comunitarias (multijugador, Steam Workshop), presencia de DLC, mes o anio de lanzamiento.
- Y (objetivo): proporcion de resenas positivas, tiempo de juego mediano, volumen de resenas.
- T (contexto temporal): juegos publicados entre 2014 y 2024.


```python
# Boceto de indicadores derivados que se construiran en el Avance 2.
# Este bloque es ilustrativo: se activara cuando el dataset este disponible
# y se confirmen los nombres de columna.

def proporcion_positivas(positivas, negativas):
    # Proporcion de resenas positivas sobre el total de resenas
    total = positivas + negativas
    if total == 0:
        return None
    return positivas / total

# Ejemplo de uso con valores de formato (no corresponden a un juego real)
print(proporcion_positivas(800, 200))
```
